EDA only. This notebook does **not** produce any reported result; all numbers in the paper come from `scripts/run_experiments.py`.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent / 'src'))
import pandas as pd
import matplotlib.pyplot as plt
from bpso_fs.data import load_dataset
```

```python
bundles = {n: load_dataset(n) for n in ['breast_cancer', 'pima_diabetes', 'heart_cleveland']}
pd.DataFrame([{k: v for k, v in b.summary().items() if k in ('name','n_samples','n_features','positive_rate','n_missing_cells')} for b in bundles.values()])
```

```python
# Missing values per feature
for n, b in bundles.items():
    m = b.X.isna().sum()
    print(n, dict(m[m > 0]))
```

```python
# Feature distributions by class (Pima)
b = bundles['pima_diabetes']
fig, axes = plt.subplots(2, 4, figsize=(12, 5))
for ax, c in zip(axes.ravel(), b.X.columns):
    for label in (0, 1):
        ax.hist(b.X.loc[b.y == label, c].dropna(), bins=25, alpha=0.5, label=f'y={label}')
    ax.set_title(c)
axes[0, 0].legend()
plt.tight_layout()
```

```python
# Correlation matrix (breast cancer) - many highly correlated, redundant features
b = bundles['breast_cancer']
plt.figure(figsize=(8, 7))
plt.imshow(b.X.corr(), cmap='RdBu_r', vmin=-1, vmax=1)
plt.colorbar()
plt.xticks(range(b.n_features), b.feature_names, rotation=90, fontsize=6)
plt.yticks(range(b.n_features), b.feature_names, fontsize=6)
plt.tight_layout()
```
